# Akaike Information Criterion


In the context of model selection we briefly touched on the Akaike Information Criterion (AIC) and used it to compare different models. Here we will go into more detail to justify its use. At the core of the AIC is the concept of information loss.

In order to understand it, we will ignore time series and focus on a very simple case of linear regression with only a slope coefficient assuming independent, normally distributed errors $$e \sim \mathcal{N}(0, \sigma^2)$$. For a single observation the model is:

$$
y = \beta x + e
$$

## Maximum Likelihood Estimation

Consider $x$ to be fixed and known. Then $y$ is normally distributed with mean $\beta x$ and variance $\sigma^2$. The probability density function of a normally distributed variable with mean $\mu$ and variance $\sigma^2$ is:

$$
f(y|\beta, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y - \mu)^2}{2\sigma^2}\right)
$$

Now replace $\mu$ with $\beta x$ to get the density of $y$ given $\beta$ and $\sigma^2$:

$$
f(y|\beta, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y - \beta x)^2}{2\sigma^2}\right)
$$

For $n$ independent observations, the likelihood function is the product of the individual densities:

$$
L(\beta, \sigma^2 | y_1, y_2, \ldots, y_n) = \prod_{i=1}^{n} f(y_i|\beta, \sigma^2) = \left(\frac{1}{\sqrt{2\pi\sigma^2}}\right)^n \exp\left(-\frac{1}{2\sigma^2} \sum_{i=1}^{n} (y_i - \beta x_i)^2\right)
$$

Usually we work with the log-likelihood function (the natural logarithm of the likelihood function) because it turns the product into a sum. We will also drop the conditioning on the data from the notation for simplicity (but keep in mind that it is still there). Further, we will assume $\sigma$ to be known for now and we will focus on $\beta$. So the log-likelihood function as a function of $\beta$ is:

$$
\mathcal{l}(\beta) = -\frac{n}{2} \log(2\pi\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{n} (y_i - \beta x_i)^2
$$


:::{.callout-important}
## ML and Mean Squared Error

Take a moment to look at this specific log-likelihood function. The first term does not depend on $\beta$. The second one involves the sum of squared errors. Remember that this was exactly the criterion we used to find the best fit line in linear regression using OLS. To maximize the log-likelihood here is equivalent to minimizing the sum of squared errors.

:::

Now we want to find what value of $\beta$ makes the log-likelihood function as large as possible. This value is called the maximum likelihood estimate (MLE) of $\beta$, denoted as $\hat{\beta}$. To find it, let's write the derivative and set it to zero:

$$
l'(\beta) = -\frac{1}{2\sigma^2} \sum_{i=1}^{n} 2(y_i - \beta x_i)(-x_i) = \frac{1}{\sigma^2} \sum_{i=1}^{n} (y_i - \beta x_i)x_i
$$

Setting the derivative to zero gives us:

$$
\sum_{i=1}^{n} (y_i - \hat{\beta} x_i)x_i = 0
$$

We will rearrange the equation to get a nice expression for $\hat{\beta}$:

$$
\sum_{i=1}^{n} y_i x_i - \hat{\beta} \sum_{i=1}^{n} x_i^2 = 0
$$

$$
\hat{\beta} = \frac{\sum_{i=1}^{n} y_i x_i}{\sum_{i=1}^{n} x_i^2}
$$



:::{.callout-important }
## Variance, Covariance, and the Estimator

Take another moment and look at the estimator $\hat{\beta}$. If $x$ and $y$ are centered
(to have zero mean), then the numerator of the estimator is the covariance between $x$ and $y$.
The denominator is the variance of $x$ (again, if $x$ is zero-centered).

:::


## Predictive Performance

With the MLE of $\beta$ we can now make predictions for new values of $y$. Say we observe a sample of new values of $y$ at the same values of $x$. These new values will come from the same distribution as the original values of $y$, however, the random errors will be *different* and *independent* of the errors in the original sample.

$$
y^{*}_i = \hat{\beta} x_i + e^{*}_i
$$

The **estimated** log-likelihood of the new sample is:

$$
\mathcal{l}(\hat{\beta}) = -\frac{n}{2} \log(2\pi\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{n} (y^{*}_i - \hat{\beta} x_i)^2
$$

Note that we are using the MLE $\hat{\beta}$ from the original sample to compute the log-likelihood of the new sample.

Now look at the second term in the estimated log-likelihood. It is simply the sum of squared residuals of the new sample. A model performs well if it can predict new data well (meaning: with a small difference between observation $y^{*}_i$ and prediction $\hat{y}^{*}_i$). In our case a small prediction error means a small sum of squared residuals and thus a large estimated log-likelihood. Therefore, if we are looking for a predictive model we should focus on the estimated log-likelihood of new data.

Now we can turn to the question behind the AIC: it asks: what is the *average* (expected) log-likelihood of new data under a model that was fitted to the original data? This is called the **expected log-likelihood**. The AIC is an estimate of this quantity. 


:::{.callout-important}
## Actual likelihood vs. estimated likelihood

In the above we have been using the estimated log-likelihood of new data. This is different from the actual log-likelihood of new data, which is:

$$
l(\beta) = -\frac{n}{2} \log(2\pi\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{n} (y^{*}_i - \beta x_i)^2
$$

We cannot compute this from data, though because it involves involves data that we have not seen and apart from that it also involves the unknown true parameter $\beta$. What we can compute is the estimated log-likelihood of the original data, which is:

$$
l(\hat{\beta}) = -\frac{n}{2} \log(2\pi\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{n} (y_i - \hat{\beta} x_i)^2
$$

and hope that it is a good estimate of the expected log-likelihood of new data. However, the next (optional) section shows that it is a biased estimate. It overestimates the expected log-likelihood of new data. The reason is that the MLE $\hat{\beta}$ was chosen to maximize the log-likelihood of the original data. Therefore, it will always perform better on the data it was fitted to than on new data. The AIC corrects for this bias.

For the linear regression model with one slope coefficient, the AIC is defined as:

$$
\text{AIC} = -2 \mathcal{l}(\hat{\beta}) + 2
$$

For a model with $k$ parameters (including the intercept and the variance), the AIC is defined as:

$$
\text{AIC} = -2 \mathcal{l}(\hat{\beta}) + 2k
$$

:::

Often the preferred criterion is a corrected version of the AIC, called the AICc. It is defined as:

$$
\text{AICc} = \text{AIC} + \frac{2k(k+1)}{n-k-1}
$$

where $n$ is the number of observations and $k$ is the number of parameters in the model. It was developed to correct for the bias of the AIC in small samples. The AICc converges to the AIC as $n$ becomes large. If in doubt, use the AICc (in some software packages you need to request it explicitly or calculate it yourself).


## Bayesian Information Criterion

Another commonly reported model selection criterion is the Bayesian Information Criterion (BIC). It is defined as:

$$
\text{BIC} = -2 \mathcal{l}(\hat{\beta}) + k \log(n)
$$

The BIC is similar to the AIC, but it penalizes models with more parameters more heavily so it favours simpler models more than the AIC does. Both criteria are constructed with a different idea in mind. The AIC is focused on predictive performance, while the BIC is focused on finding the true model.
