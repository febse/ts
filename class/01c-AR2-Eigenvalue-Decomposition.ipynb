# AR(2): Eigenvalue Decomposition

:::{.ex-class}
$$
x_t = \phi_1 x_{t - 1} + \phi_2 x_{t - 2}
$$
:::

:::{.ex-class}
$$
\begin{align*}
x_t & = \phi_1 x_{t - 1} + \phi_2 x_{t - 2}\\
x_{t - 1} & = x_{t - 1}\\
\end{align*}
$$
:::
